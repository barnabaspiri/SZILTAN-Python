##### Feladat (Példatár 5.2)

Az ábrán látható $AB$ és $BC$ rúd csuklósan kapcsolódik a $B$ pontban. A tartó terhelése a $P$ koncentrált erő és az állandó intenzitású $q$ megoszló terhelés. Határozzuk meg a $B$ csukló lehajlását a szuperpozíció elvének és a járulékképletek felhasználásával!

Adatok: $P$, $q$, $a$, $b$, $IE$ (szimbolikusan).

Feladatok:

a) A $B$ csukló lehajlása ($f_B$) szuperpozícióval?

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

A feladat szimbolikusan adott, numerikus adatok nincsenek.

```python
P, q, a, b, IE = sp.symbols('P, q, a, b, IE', positive=True)
```

#### a) A $B$ csukló lehajlása szuperpozícióval

##### Reakciók

<img src="img/02.jpg" alt="Feladat" width=500px>

A $B$ csuklóban a hajlítónyomaték zérus, így a nyomatéki egyenleteket a csukló bal ($M_{h1}$) és jobb ($M_{h2}$) oldaláról is felírhatjuk.

```python
F_A, F_C, M_C = sp.symbols('F_A, F_C, M_C') # Ismeretlen reakciók

# y irányú erő egyensúly
eq1 = F_A - P - q*b + F_C
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# Hajlítónyomaték a B csuklóban (bal oldalról)
eq2 = -F_A*a + P*a/3
display(Math(rf'M_{{h,B}} = M_{{h1}}(x = a) = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))

# Hajlítónyomaték a B csuklóban (jobb oldalról)
eq3 = -F_C*b + q*b**2/2 + M_C
display(Math(rf'M_{{h,B}} = M_{{h2}}(x = a) = 0 \quad \longrightarrow \quad {latex(eq3)} = 0'))
```

```python
# Rendszer megoldása
sol = sp.solve([eq1, eq2, eq3], [F_A, F_C, M_C])

F_A_sol = sol[F_A]
display(Math(rf'F_A = {latex(F_A_sol)}'))

F_C_sol = sol[F_C]
display(Math(rf'F_C = {latex(F_C_sol)}'))

M_C_sol = sp.factor(sol[M_C])
display(Math(rf'M_C = {latex(M_C_sol)}'))
```

##### $B$ csuklón átadódó erő

<img src="img/03.jpg" alt="Feladat" width=350px>

```python
# Az AB rúd függőleges egyensúlyából
F_B = P - F_A_sol
display(Math(rf'F_B = P - F_A = {latex(F_B)}'))
```

##### A $BC$ rúd deformációja 2 terhelésből adódik

<img src="img/04.jpg" alt="Feladat" width=650px>

Szuperpozíció elve:

$$f_B = f_{B1} + f_{B2}$$

ahol:
- $f_{B1}$: a $q$ megoszló terhelésből adódó lehajlás (I. eset),
- $f_{B2}$: a $B$ csuklón átadódó $F_B$ erőből adódó lehajlás (II. eset).

##### Járulékképletek (képletgyűjteményből)

Befogott tartó szabad végének lehajlása:
- állandó intenzitású megoszló terhelés esetén: $f = \frac{q \cdot b^4}{8 IE}$,
- a szabad végen működő koncentrált erő esetén: $f = \frac{F \cdot b^3}{3 IE}$.

```python
# I. eset: megoszló terhelés
f_B1 = q * b**4 / (8 * IE)
display(Math(rf'f_{{B1}} = {latex(f_B1)}'))

# II. eset: koncentrált erő a szabad végen
f_B2 = F_B * b**3 / (3 * IE)
display(Math(rf'f_{{B2}} = \frac{{F_B \cdot b^3}}{{3 IE}} = {latex(f_B2)}'))
```

```python
# Szuperpozíció
f_B = f_B1 + f_B2
display(Math(rf'f_B = f_{{B1}} + f_{{B2}} = {latex(f_B)}'))
```

##### Ellenőrzés a rugalmas szál differenciálegyenletével

A $BC$ rudat befogott tartóként vizsgáljuk, amelyet a $q$ megoszló terhelés és a $B$ pontban lefelé mutató $F_B$ erő terhel. Az $x$ tengely a $B$ pontból indul és jobbra mutat ($x \in [0, b]$), a hajlítónyomatékot a $B$ pont felőli oldalról írjuk fel.

$$-IE \cdot w''(x) = M_h(x)$$

Peremfeltételek a $C$ befogásnál: $w(b) = 0$ és $w'(b) = 0$.

```python
x = sp.Symbol('x')
C_1, C_2 = sp.symbols('C_1, C_2') # Integrálási konstansok

# Hajlítónyomatéki függvény
Mh = F_B*x + q*x**2/2
display(Math(rf'M_h(x) = F_B \cdot x + \frac{{q \cdot x^2}}{{2}} = {latex(Mh)}'))

# Első integrálás
int_1 = sp.integrate(Mh, x) + C_1
display(Math(rf'-IE \cdot w^{{\prime}}(x) = \int M_h(x) \, dx = {latex(int_1)}'))

# Második integrálás
int_int = sp.integrate(int_1, x) + C_2
display(Math(rf'-IE \cdot w(x) = \iint M_h(x) \, dx = {latex(int_int)}'))
```

```python
# Peremfeltételek (a -IE szorzó kiesik)
eq1 = int_1.subs(x, b)
display(Math(rf'w^{{\prime}}(b) = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

eq2 = int_int.subs(x, b)
display(Math(rf'w(b) = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))

# Rendszer megoldása
sol = sp.solve([eq1, eq2], [C_1, C_2])
display(Math(rf'C_1 = {latex(sol[C_1])}, \qquad C_2 = {latex(sol[C_2])}'))
```

```python
# Lehajlásfüggvény
w = (-int_int / IE).subs(sol)

# A B pont lehajlása (lefelé pozitív)
f_B_ell = sp.expand(-w.subs(x, 0))
display(Math(rf'f_B = -w(0) = {latex(f_B_ell)}'))

# Összehasonlítás a szuperpozícióval kapott eredménnyel
display(Math(rf'f_B - f_B^{{\text{{szuperpozíció}}}} = {latex(sp.simplify(f_B_ell - f_B))}'))
```

A két módszer ugyanazt az eredményt adja, a szuperpozícióval és a járulékképletekkel jóval gyorsabban jutunk el a $B$ csukló lehajlásához.
