##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
a_1 = 20 # mm
b_1 = 10 # mm
a_2 = 10 # mm
b_2 = 20 # mm
```

##### Részekre bontás

<img src="img/02.jpg" alt="Feladat" width=300px>

##### Az adott részek tulajdonságai

```python
# 1. téglalap
S1xi = a_1/2
S1eta = (b_2 - b_1) + b_1/2

A_1 = a_1 * b_1

# 2. téglalap
S2xi = a_1 + a_2/2
S2eta = b_2/2

A_2 = a_2 * b_2
```

##### Teljes keresztmetszet súlypontja

```python
Sxi = (S1xi*A_1 + S2xi*A_2) / (A_1 + A_2)
display(Math(rf'S_{{\xi}} = {latex(Sxi)} ~\mathrm{{mm}}'))

Seta = (S1eta*A_1 + S2eta*A_2) / (A_1 + A_2)
display(Math(rf'S_{{\eta}} = {latex(Seta)} ~\mathrm{{mm}}'))
```

##### Másodrendű nyomatékok az alakzatok saját tengelyeire

```python
# 1. téglalap
I_x1 = a_1*b_1**3 / 12
I_y1 = a_1**3*b_1 / 12
I_x1y1 = 0

# 2. téglalap
I_x2 = a_2*b_2**3 / 12
I_y2 = a_2**3*b_2 / 12
I_x2y2 = 0
```

##### Másodrendű nyomatékok a $\xi$ és $\eta$ tengelyekre a Steiner-tétel segítségével

```python
I_xi = (I_x1 + S1eta**2*A_1) + (I_x2 + S2eta**2*A_2)
display(Math(rf'I_{{\xi}} = {latex(round(I_xi, 2))} ~\mathrm{{mm^4}}'))

I_eta = (I_y1 + S1xi**2*A_1) + (I_y2 + S2xi**2*A_2)
display(Math(rf'I_{{\eta}} = {latex(round(I_eta, 2))} ~\mathrm{{mm^4}}'))

I_xieta = (I_x1y1 + (-S1xi)*(-S1eta)*A_1) + (I_x2y2 + (-S2xi)*(-S2eta)*A_2)
display(Math(rf'I_{{\xi\eta}} = {latex(round(I_xieta, 2))} ~\mathrm{{mm^4}}'))
```

##### Átszámítás a síkidom súlyponti tengelyeire

```python
# A síkidom teljes területe
A = A_1 + A_2

# Átszámítás az x tengelyre
# I_xi = I_x + Seta**2*A átrendezve:
I_x = I_xi - Seta**2*A
display(Math(rf'I_{{x}} = {latex(round(I_x, 2))} ~\mathrm{{mm^4}}'))

# Átszámítás az y tengelyre
# I_eta = I_y + Sxi**2*A átrendezve:
I_y = I_eta - Sxi**2*A
display(Math(rf'I_{{y}} = {latex(round(I_y, 2))} ~\mathrm{{mm^4}}'))

# Átszámítás az xy tengelypárra
# I_xieta = I_xy + Seta*Sxi*A átrendezve:
I_xy = I_xieta -  Seta*Sxi*A
display(Math(rf'I_{{xy}} = {latex(round(I_xy, 2))} ~\mathrm{{mm^4}}'))
```

##### Főmásodrendű nyomatékok

```python
I_1 = (I_x + I_y) / 2 + 1/2 * sp.sqrt( (I_x - I_y)**2 + 4*I_xy**2 )
display(Math(rf'I_{{1}} = {latex(round(I_1, 2))} ~\mathrm{{mm^4}}'))

I_2 = (I_x + I_y) / 2 - 1/2 * sp.sqrt( (I_x - I_y)**2 + 4*I_xy**2 )
display(Math(rf'I_{{2}} = {latex(round(I_2, 2))} ~\mathrm{{mm^4}}'))
```

##### Főirányok

```python
# Az x tengely és az 1. főmásodrendű nyomatéki tengely által bezárt szög
alpha = sp.atan((I_x - I_1) / (I_xy))
alpha = sp.deg(alpha)
display(Math(rf'\alpha = {latex(round(alpha, 2))}^\circ'))
```
