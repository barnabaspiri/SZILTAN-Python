## 2.1. Másodrendű nyomatékok a súlyponti tengelyekre

Súlypont számítása:

<img src="fig/fig02.png" style="width: 400px;">

```python
# Import
import numpy as np
```

```python
# Adatok
a1 = 2
b1 = 1
a2 = 1
b2 = 2
```

```python
# Az egyszerű alakzatok (négyszögek) súlypontjainak koordinátái
S1ξ = 1.0
S1η = 1.5
S2ξ = 2.5
S2η = 1.0

# A négyszögek területe
A1 = 2.0
A2 = 2.0
```

```python
# Az összetett keresztmetszet súlypontja
ξ_s = (S1ξ * A1 + S2ξ * A2) / (A1 + A2)
η_s = (S1η * A1 + S2η * A2) / (A1 + A2)

print(f'Súlypont (ξ,η): \nξ_s = {ξ_s:.2f} cm\nη_s = {η_s:.2f} cm')
```

```python
# Másodrendű nyomatékok az alakzatok saját súlyponti tengelyeire
I_x1 = a1 * b1**3 / 12
I_y1 = a1**3 * b1 / 12
I_xy1 = 0

I_x2 = a2 * b2**3 / 12
I_y2 = a2**3 * b2 / 12
I_xy2 = 0
```

```python
# Másodrendű nyomatékok a Steiner-tétel segítségével (ξ, η) tengelyekre
I_ξ1 = I_x1 + S1η**2 * A1
I_ξ2 = I_x2 + S2η**2 * A2
I_ξ = I_ξ1 + I_ξ2

I_η1 = I_x1 + S1ξ**2 * A1
I_η2 = I_x2 + S2ξ**2 * A2
I_η = I_η1 + I_η2

I_ξη1 = I_xy1 + (-S1ξ) * (-S1η) * A1
I_ξη2 = I_xy2 + (-S2ξ) * (-S2η) * A2
I_ξη = I_ξη1 + I_ξη2

print(f'Másodrendű nyomatékok (ξ,η): \nI_ξ = {I_ξ:.4f} cm^4\nI_η = {I_η:.4f} cm^4\nI_ξη = {I_ξη:.4f} cm^4')
```

```python
# Átszámítás a súlyponti tengelyekre
I_x = I_ξ - η_s**2 * (A1 + A2)
I_y = I_η - ξ_s**2 * (A1 + A2)
I_xy = I_ξη - η_s * ξ_s * (A1 + A2)

print(f'Másodrendű nyomatékok (x,y): \nI_x = {I_x:.4f} cm^4\nI_y = {I_y:.4f} cm^4\nI_xy = {I_xy:.4f} cm^4')
```

## 2.2. Főmásodrendű nyomatékok és főirányok

```python
I1 = (I_x + I_y) / 2 + 0.5 * np.sqrt( (I_x - I_y)**2 + 4*I_xy**2)
I2 = (I_x + I_y) / 2 - 0.5 * np.sqrt( (I_x - I_y)**2 + 4*I_xy**2)

α1 = np.atan( (I_x - I1) / (I_xy) )
α2 = α1 + np.pi/2

α1 = np.rad2deg(α1)
α2 = np.rad2deg(α2)

print(f'Főmásodrendű nyomatékok: \nI1 = {I1:.4f} cm^4\nI2 = {I2:.4f} cm^4\n')
print(f'Főirányok: \nα1 = {α1:.4f} deg,\nα2 = {α2:.4f} deg')
```
