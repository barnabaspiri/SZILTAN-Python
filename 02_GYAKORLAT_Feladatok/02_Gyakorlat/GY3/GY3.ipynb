A poláris másodrendű nyomaték a következőképpen számítható:

$$ I_p = \int\limits_{(A)} \rho^2 \,\mathrm{d}A = \int\limits_{0}^{2\pi}\int\limits_{0}^{r} \rho^3 \, \mathrm{d}\rho \mathrm{d}\varphi$$

```python
import sympy as sp
```

```python
rho, phi, r, d = sp.symbols('ρ, φ, r, d')

I_p = sp.integrate( sp.integrate( rho**3, (rho, 0, r) ), (phi, 0, 2*sp.pi) )
print(f'Poláris másodrendű nyomaték: I_p')
display(I_p)

print(f'\nPoláris másodrendű nyomaték (átmérővel): I_p')
I_p = I_p.subs(r,d/2)
display(I_p)
print('')
```

A körszimmetria miatt:

$$ I_x = I_y $$ és $$ I_{xy} = 0.$$

Emellett
$$ I_p = I_x + I_y = 2 I_x \quad \longrightarrow \quad I_x = I_p / 2 $$

```python
print(f'\nMásodrendű nyomaték (átmérővel): I_x = I_y')
I_x = I_p/2
I_y = I_x
display(I_x)
```
