##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
h = 120 # mm
b = 120 # mm
s = 5 # mm
A = 17.1781 # cm^2
e_x1 = 4.4033 # cm
I_X = 441.2373 # cm^4
I_Y = 264.1328 # cm^4
k = 20 # mm
d = 30 # mm

# Mindent mm átváltása cm-be
h = h/10
b = b/10
s = s/10
k = k/10
d = d/10
```

##### Részekre bontás

<img src="img/02.jpg" alt="Feladat" width=250px>

##### Az egyes részek tulajdonságai

```python
# U szelvény
S_1 = sp.Matrix([e_x1, h/2])
A_1 = A

# Furatgyengítés
S_2 = sp.Matrix([s/2, h/2+k])
A_2 = d*s
```

##### A furatgyengített keresztmetszet súlypontja

```python
xi_S = (S_1[0]*A_1 - S_2[0]*A_2) / (A_1 - A_2)
display(Math(rf'\xi_{{S}} = {latex(round(xi_S,4))} ~ \mathrm{{cm}}'))

eta_S = (S_1[1]*A_1 - S_2[1]*A_2) / (A_1 - A_2)
display(Math(rf'\eta_{{S}} = {latex(round(eta_S,4))} ~ \mathrm{{cm}}'))
```

##### Az eredeti $X-Y$ és az új $x-y$ tengelypárok távolsága

```python
x_s = xi_S - e_x1
display(Math(rf'x_{{s}} = {latex(round(x_s,4))} ~ \mathrm{{cm}}'))

y_s = h/2 - eta_S
display(Math(rf'y_{{s}} = {latex(round(y_s,4))} ~ \mathrm{{cm}}'))
```

##### Másodrendű nyomatékok

```python
# U szelvény
I_x1 = I_X
I_y1 = I_Y
I_x1y1 = 0

# Furatgyengítés
I_x2 = (s*d**3)/12
I_y2 = (s**3*d)/12
I_x2y2 = 0

# x tengelyre
I_x = (I_x1 + y_s**2*A_1) - (I_x2 + (y_s+k)**2*A_2)
display(Math(rf'I_{{x}} = {latex(round(I_x,4))} ~ \mathrm{{cm^4}}'))

# y tengelyre
I_y = (I_y1 + x_s**2*A_1) - (I_y2 + (x_s+e_x1-s/2)**2*A_2)
display(Math(rf'I_{{y}} = {latex(round(I_y,4))} ~ \mathrm{{cm^4}}'))

# xy tengelypárra
I_xy = (I_x1y1 + (x_s)*(-y_s)*A_1) - (I_x2y2 + (x_s+e_x1-s/2)*(-(k+y_s))*A_2)
display(Math(rf'I_{{xy}} = {latex(round(I_xy,4))} ~ \mathrm{{cm^4}}'))
```

##### Főmásodrendű nyomatékok

```python
# 1. főmásodrendű nyomaték
I_1 = (I_x + I_y) / 2 + 1/2 * sp.sqrt( (I_x - I_y)**2 + 4*I_xy**2 )
display(Math(rf'I_{{1}} = {latex(round(I_1,4))} ~ \mathrm{{cm^4}}'))

# 2. főmásodrendű nyomaték
I_2 = (I_x + I_y) / 2 - 1/2 * sp.sqrt( (I_x - I_y)**2 + 4*I_xy**2 )
display(Math(rf'I_{{2}} = {latex(round(I_2,4))} ~ \mathrm{{cm^4}}'))

# az x és az 1. főmásodrendű nyomatéki tengely által bezárt szög
alpha = sp.atan( (I_1-I_x) / (I_xy))
alpha = sp.deg(alpha)
display(Math(rf'\alpha = {latex(round(alpha,2))}^\circ'))
```
