# 2. Megoldás

```python
import sympy as sp
```

Elemi kis darab vizsgálata:

<img src="fig/fig02.png" style="width: 300px;">

```python
a, xi, eta, m = sp.symbols('a, ξ, η, m')
```

Másodrendú nyomaték a $\xi$ tengelyre:
    
$$ I_\xi = \int\limits_{(A)} \eta^2 \,\mathrm{d}A = \int\limits_{0}^{m} \eta^2 v(\eta)\,\mathrm{d}\eta = \int\limits_{0}^{m} \eta^2 (a-\frac{a}{m}\eta)\,\mathrm{d}\eta $$

```python
v = a - a / m * eta
I_xi = sp.integrate( eta**2 * v, (eta, 0, m) )
print('Másodrendű nyomaték: I_ξ')
display(I_xi)
```

```python
t = m/3
A = a*m/2

I_x = I_xi - t**2*A

print('Másodrendű nyomaték: I_x')
display(I_x)
```

A tengelypárra számolt másodrendű nyomaték:

$$ I_{\xi\eta} = \int\limits_{(A)}\xi \eta \, \mathrm{d}A, $$

amelynek az eredménye:

$$ I_{\xi\eta} = \pm \frac{abm^2}{12} $$

és

$$ I_{xy} = \pm \frac{abm^2}{36}, $$

melynek előjele attól függ, hogy a háromszög csúcsa melyik térnegyedben helyezkedik el:
