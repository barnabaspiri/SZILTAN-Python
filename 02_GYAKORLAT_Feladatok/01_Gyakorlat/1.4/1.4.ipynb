##### Feladat (Példatár 1.7)

A vízszintes, merevnek tekintett rudat két huzal tartja egyensúlyban, melyek távolsága egymástól $L = 1 ~\mathrm{m}$ . A rúd terhelése a rajta elhelyezett $G = 5000 ~\mathrm{N}$  súlyú teher. A bal oldali huzal anyaga acél, rugalmassági modulusza $E_1 = 200 ~\mathrm{GPa}$ . A jobb oldali huzal alumíniumból készült, rugalmassági modulusza $E_2 = 70 ~\mathrm{GPa}$ . A huzalok keresztmetsze-tei és hosszaik az alábbiak: $A_1 = 120 ~\mathrm{mm^2}$ , $A_2 = 240 ~\mathrm{mm^2}$ , $L_1 = 15 ~\mathrm{m}$ , $L_2 = 25 ~\mathrm{m}$ .

Hol helyezkedjen el a $G$ teher, ha azt szeretnénk, hogy a merev rúd a terhelés hatására vízszintes maradjon?

Mekkora feszültségek ébrednek ekkor a huzalokban? A vízszintes gerenda súlyát elhanyagoljuk!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L = 1 # m
G = 5000 # N
E_1 = 200 # GPa
E_2 = 70 # GPa
A_1 = 120 # mm^2
A_2 = 240 # mm^2
L_1 = 15 # m
L_2 = 25 # m

# Átváltás N, mm, MPa rendszerbe
L = L*10**3 # mm
L_1 = L_1*10**3 # mm
L_2 = L_2*10**3 # mm
E_1 = E_1*10**3 # MPa
E_2 = E_2*10**3 # MPa
```

##### Ha a rúd vízszintes, akkor a két kötéls megynyúlása megegyezik: $\Delta L_1 = \Delta L_2$

```python
# Szimbólikus változók
N_1, N_2 = sp.symbols('N_1, N_2')

# Megnyúlások megegyeznek
delta_L1 = (N_1*L_1) / (A_1*E_1)
delta_L2 = (N_2*L_2) / (A_2*E_2)

eq1 = delta_L1 - delta_L2

# A súlyerő megegyezik a normálerők összegével
eq2 = N_1 + N_2 - G

# Rendszer megoldása
sol = sp.solve([eq1, eq2], (N_1, N_2))
N_1 = sol[N_1]
N_2 = sol[N_2]

display(Math(rf'N_1 = {latex(N_1.evalf(6))} ~\mathrm{{N}}'))
display(Math(rf'N_2 = {latex(N_2.evalf(6))} ~\mathrm{{N}}'))
```

##### Nyomatéki egyensúly az $A$ pontra

```python
# Szimbólikus változó
a = sp.Symbol('a')

# Nyomatéki egyensúly
eq = -G*a + N_2*L

asol = sp.solve(eq, a)[0]
display(Math(rf'a = {latex(round(asol, 3))} ~\mathrm{{mm}}'))
```
