##### Feladat (Példatár 1.1)

Egy $G = 700 ~\mathrm{N}$  súlyú lámpatestet szeretnék felfüggeszteni a merevnek tekintett falak közé egy $d = 3 ~\mathrm{mm}$  átmérőjű acélhuzallal, melyre húzás esetén a megengedett feszültség $\sigma_\mathrm{meg} = 285 ~\mathrm{MPa}$ . Határozzuk meg, hogy mekkora lehet minimálisan a huzal vízszintessel bezárt szöge annak érdekében, hogy a huzalban ébredő normálfeszültség a megengedett érték alatt maradjon!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
d = 3 # mm
sigma_meg = 285 # MPa
G = 700 # N
```

##### Szabadtest ábra

<img src="img/02.jpg" alt="Feladat" width=200px>

##### Ismeretlen erővektor

```python
F, alpha = sp.symbols('F, α')
```

##### Erőegyensúly

```python
# x irány
eq1 = -F*sp.cos(alpha) + F*sp.cos(alpha)

# y irány
eq2 = -G + 2*F*sp.sin(alpha)
```

##### Az $y$ komponensből fejezzük ki $F$-et

```python
sol = sp.solve(eq2, F)
Fsol = sol[0]
display(Math(rf'F = {latex(Fsol.evalf(6))}'))
```

##### A kötél megfelel, ha $\sigma \leq \sigma_\mathrm{meg}$, határhelyzet az egyenlőség esete

```python
# A kötél keresztmetszete
A = d**2*sp.pi/4

# Az ébredő feszültség F hatására
sigma = Fsol / A

# Határhelyzet
eq = sigma - sigma_meg

# Egyenlet megoldása az ismeretlen szögre
sol = sp.solveset(eq, alpha, domain=sp.Interval(0, sp.pi/2))
alphasol = list(sol)[0]
alphasol = sp.deg(alphasol)
display(Math(rf'\alpha = {latex(alphasol.evalf(4))}^\circ'))
```
