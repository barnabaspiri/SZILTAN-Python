##### Feladat (Példatár 3.3)

Egy acélszerkezetet alkotó I szelvény $P$ pontjába nyúlásmérő bélyeget ragasztunk és mérjük az alakváltozásokat $a$, $b$ és $c$ irányokban. Határozzuk meg a feszültségi állapotot $P$-ben.

Adatok: $\varepsilon_a = 100 \cdot 10^{-6}$, $\varepsilon_b = 50 \cdot 10^{-6}$, $\varepsilon_c = -70 \cdot 10^{-6}$, $\alpha_a = 0^\circ$, $\alpha_b = 70^\circ$, $\alpha_c = 200^\circ$, $E = 210 ~\mathrm{GPa}$, $\nu = 0.3$.

Feladatok:

a) Feszültségi állapot?

b) Főfeszültségek és főnyúlások?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
eps_a = 100e-6
eps_b = 50e-6
eps_c = -70e-6
alpha_a = 0 # °
alpha_b = 70 # °
alpha_c = 200 # °
E = 210 # GPa
nu = 0.3

# Átváltás N, mm, MPa rendszerbe
E = E * 10**3 # MPa
```

#### a) Feszültségi állapot

##### $a$, $b$, $c$ irányba mutató egységvektorok

Az $\alpha_a$, $\alpha_b$, $\alpha_c$ szögeket az $x$ tengelytől mérjük, pozitív forgásirányban.

```python
# Az x-y síkba eső, az x tengellyel α szöget bezáró egységvektor
def n(alpha):
    return sp.Matrix([sp.cos(alpha), sp.sin(alpha), 0])

# Szimbólikus szögek a kiíratáshoz
alpha_a_s, alpha_b_s, alpha_c_s = sp.symbols('alpha_a, alpha_b, alpha_c')

n_a = n(sp.rad(alpha_a))
n_b = n(sp.rad(alpha_b))
n_c = n(sp.rad(alpha_c))

display(Math(rf'\mathbf{{n}}_a = {latex(n(alpha_a_s))} = {latex(n_a.evalf(3))}'))
display(Math(rf'\mathbf{{n}}_b = {latex(n(alpha_b_s))} = {latex(n_b.evalf(3))}'))
display(Math(rf'\mathbf{{n}}_c = {latex(n(alpha_c_s))} = {latex(n_c.evalf(3))}'))
```

##### Alakváltozási tenzor

```python
eps_x, eps_y, eps_z = sp.symbols('varepsilon_x, varepsilon_y, varepsilon_z') # Fajlagos nyúlások
gamma_xy, gamma_xz, gamma_yz = sp.symbols('gamma_xy, gamma_xz, gamma_yz') # Fajlagos szögváltozások

eps = sp.Matrix([[     eps_x, gamma_xy/2, gamma_xz/2],
                 [gamma_xy/2,      eps_y, gamma_yz/2],
                 [gamma_xz/2, gamma_yz/2,      eps_z]])
display(Math(rf'\boldsymbol{{\varepsilon}} = {latex(eps)}'))
```

##### Fajlagos nyúlás egy adott irányba

$$\varepsilon_p = \left( \boldsymbol{\varepsilon} \cdot \mathbf{n}_p \right) \cdot \mathbf{n}_p, \qquad \text{ahol} \quad \mathbf{n}_p = \begin{bmatrix} p_1 \\ p_2 \\ p_3 \end{bmatrix}$$

```python
p_1, p_2, p_3 = sp.symbols('p_1, p_2, p_3')
n_p = sp.Matrix([p_1, p_2, p_3])

eps_p = (eps * n_p).dot(n_p)
display(Math(rf'\varepsilon_p = \left( \boldsymbol{{\varepsilon}} \cdot \mathbf{{n}}_p \right) \cdot \mathbf{{n}}_p = {latex(sp.expand(eps_p))}'))
```

##### A nyúlások az $a$, $b$, $c$ irányokban

```python
# a irány
eps_a_n = sp.expand((eps * n_a).dot(n_a)).evalf()
eq_1 = eps_a_n - eps_a
display(Math(rf'\varepsilon_a = \left( \boldsymbol{{\varepsilon}} \cdot \mathbf{{n}}_a \right) \cdot \mathbf{{n}}_a \quad \rightarrow \quad {latex(eps_a)} = {latex(eps_a_n.evalf(3))} \qquad (1)'))

# b irány
eps_b_n = sp.expand((eps * n_b).dot(n_b)).evalf()
eq_2 = eps_b_n - eps_b
display(Math(rf'\varepsilon_b = \left( \boldsymbol{{\varepsilon}} \cdot \mathbf{{n}}_b \right) \cdot \mathbf{{n}}_b \quad \rightarrow \quad {latex(eps_b)} = {latex(eps_b_n.evalf(3))} \qquad (2)'))

# c irány
eps_c_n = sp.expand((eps * n_c).dot(n_c)).evalf()
eq_3 = eps_c_n - eps_c
display(Math(rf'\varepsilon_c = \left( \boldsymbol{{\varepsilon}} \cdot \mathbf{{n}}_c \right) \cdot \mathbf{{n}}_c \quad \rightarrow \quad {latex(eps_c)} = {latex(eps_c_n.evalf(3))} \qquad (3)'))
```

##### (1), (2), (3) egyenletrendszer megoldása

```python
# Rendszer megoldása
sol = sp.solve([eq_1, eq_2, eq_3], (eps_x, eps_y, gamma_xy))
eps_x_sol = sol[eps_x]
eps_y_sol = sol[eps_y]
gamma_xy_sol = sol[gamma_xy]

display(Math(rf'\varepsilon_x = {latex((eps_x_sol * 10**6).evalf(5))} \cdot 10^{{-6}}'))
display(Math(rf'\varepsilon_y = {latex((eps_y_sol * 10**6).evalf(5))} \cdot 10^{{-6}}'))
display(Math(rf'\gamma_{{xy}} = {latex((gamma_xy_sol * 10**6).evalf(5))} \cdot 10^{{-6}}'))
```

##### Síkfeszültségi állapot

A $P$ pont a szelvény terheletlen felületén van, ezért $\sigma_z = \tau_{xz} = \tau_{yz} = 0$ (síkfeszültségi állapot). Így $\gamma_{xz} = \gamma_{yz} = 0$, az $\varepsilon_z$ pedig nem független, a $\sigma_z = 0$ feltételből számítható:

$$\varepsilon_z = -\frac{\nu}{1-\nu} \left( \varepsilon_x + \varepsilon_y \right)$$

```python
eps_z_sol = -nu / (1 - nu) * (eps_x_sol + eps_y_sol)
display(Math(rf'\varepsilon_z = -\frac{{\nu}}{{1-\nu}} \left( \varepsilon_x + \varepsilon_y \right) = {latex((eps_z_sol * 10**6).evalf(5))} \cdot 10^{{-6}}'))
```

##### Így az alakváltozási tenzor

```python
eps_sol = eps.subs({eps_x: eps_x_sol, eps_y: eps_y_sol, eps_z: eps_z_sol,
                    gamma_xy: gamma_xy_sol, gamma_xz: 0, gamma_yz: 0})
display(Math(rf'\boldsymbol{{\varepsilon}} = {latex((eps_sol * 10**6).evalf(5))} \cdot 10^{{-6}}'))
```

##### Általános Hooke törvény

$$\boldsymbol{\sigma} = \frac{E}{1+\nu} \left( \boldsymbol{\varepsilon} + \frac{\nu}{1-2\nu} \cdot \varepsilon_I \cdot \mathbf{E} \right)$$

ahol:
- $\boldsymbol{\sigma}$: a feszültségi tenzor,
- $\boldsymbol{\varepsilon}$: az alakváltozási tenzor,
- $E$: a rugalmassági modulusz,
- $\nu$: a Poisson-tényező,
- $\varepsilon_I = \mathrm{tr}(\boldsymbol{\varepsilon})$: az alakváltozási tenzor első skalárinvariánsa (a főátlóbeli elemek összege),
- $\mathbf{E}$: az egységmátrix.

```python
I = sp.eye(3) # Egységmátrix

# Az alakváltozási tenzor első skalárinvariánsa
eps_I = eps_sol.trace()
display(Math(rf'\varepsilon_I = \mathrm{{tr}}(\boldsymbol{{\varepsilon}}) = \varepsilon_x + \varepsilon_y + \varepsilon_z = {latex((eps_I * 10**6).evalf(5))} \cdot 10^{{-6}}'))
```

```python
sigma = E / (1 + nu) * (eps_sol + nu / (1 - 2*nu) * eps_I * I)

# A kerekítési hibából adódó, nullához közeli értékek nullázása
sigma = sigma.evalf(chop=1e-10)

# Például a σ_x feszültség
display(Math(rf'\sigma_x = \frac{{E}}{{1+\nu}} \left( \varepsilon_x + \frac{{\nu}}{{1-2\nu}} \cdot \varepsilon_I \right) = {latex(round(sigma[0, 0], 3))} ~\mathrm{{MPa}}'))
```

##### Így a feszültségi állapot

```python
display(Math(rf'\boldsymbol{{\sigma}} = {latex(sigma.evalf(5))} ~\mathrm{{MPa}}'))
```

#### b) Főfeszültségek és főnyúlások

##### Főfeszültségek

$$\det \left( \boldsymbol{\sigma} - \sigma \cdot \mathbf{E} \right) = 0$$

```python
s = sp.Symbol('sigma') # Ismeretlen főfeszültség

display(Math(rf'\boldsymbol{{\sigma}} - \sigma \cdot \mathbf{{E}} = {latex((sigma - s*I).evalf(5))}'))

# Determináns kifejtése
char_eq = (sigma - s*I).det()
display(Math(rf'\det \left( \boldsymbol{{\sigma}} - \sigma \cdot \mathbf{{E}} \right) = {latex(char_eq.evalf(5))} = 0'))
```

```python
# A karakterisztikus egyenlet gyökei
sol = sp.roots(char_eq, s, multiple=True)

# Sorbarendezés után: σ_1 >= σ_2 >= σ_3
sigma_1, sigma_2, sigma_3 = sorted(sol, reverse=True)

display(Math(rf'\sigma_1 = {latex(round(sigma_1, 3))} ~\mathrm{{MPa}}'))
display(Math(rf'\sigma_2 = {latex(round(sigma_2, 3))} ~\mathrm{{MPa}}'))
display(Math(rf'\sigma_3 = {latex(round(sigma_3, 3))} ~\mathrm{{MPa}}'))
```

##### Főnyúlások

$$\det \left( \boldsymbol{\varepsilon} - \varepsilon \cdot \mathbf{E} \right) = 0$$

```python
e = sp.Symbol('varepsilon') # Ismeretlen főnyúlás

display(Math(rf'\boldsymbol{{\varepsilon}} - \varepsilon \cdot \mathbf{{E}} = {latex((eps_sol - e*I).evalf(5))}'))

# Determináns kifejtése
char_eq = (eps_sol - e*I).det()
display(Math(rf'\det \left( \boldsymbol{{\varepsilon}} - \varepsilon \cdot \mathbf{{E}} \right) = {latex(char_eq.evalf(5))} = 0'))
```

```python
# A karakterisztikus egyenlet gyökei
sol = sp.roots(char_eq, e, multiple=True)

# Sorbarendezés után: ε_1 >= ε_2 >= ε_3
eps_1, eps_2, eps_3 = sorted(sol, reverse=True)

display(Math(rf'\varepsilon_1 = {latex((eps_1 * 10**6).evalf(5))} \cdot 10^{{-6}}'))
display(Math(rf'\varepsilon_2 = {latex((eps_2 * 10**6).evalf(5))} \cdot 10^{{-6}}'))
display(Math(rf'\varepsilon_3 = {latex((eps_3 * 10**6).evalf(5))} \cdot 10^{{-6}}'))
```

##### Ellenőrzés beépített függvénnyel

```python
# Szimmetrikus mátrix sajátértékei (növekvő sorrendben)
sajatertekek_sigma = np.linalg.eigvalsh(np.array(sigma, dtype=float))
sajatertekek_eps = np.linalg.eigvalsh(np.array(eps_sol, dtype=float))

# Sorbarendezés csökkenő sorrendbe
sajatertekek_sigma = sajatertekek_sigma[::-1]
sajatertekek_eps = sajatertekek_eps[::-1]

display(Math(rf'\left[ \sigma_1 ~ \sigma_2 ~ \sigma_3 \right] = {latex(sp.Matrix(np.round(sajatertekek_sigma, 3) + 0.0).T)} ~\mathrm{{MPa}}'))
display(Math(rf'\left[ \varepsilon_1 ~ \varepsilon_2 ~ \varepsilon_3 \right] = {latex(sp.Matrix(np.round(sajatertekek_eps * 10**6, 2) + 0.0).T)} \cdot 10^{{-6}}'))
```
