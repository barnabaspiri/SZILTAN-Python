##### Feladat (Példatár 3.2)

Egy $L$ hosszúságú egyenes, $d$ átmérőjű kör keresztmetszetű, rugalmas rudat a végein $10 ~\mathrm{kN}$ nagyságú nyomóerők terhelik.

Adatok: $F = 10 ~\mathrm{kN}$, $L = 160 ~\mathrm{mm}$, $d = 79.8 ~\mathrm{mm}$, $E = 200 ~\mathrm{GPa}$, $\nu = 0.3$.

Feladatok:

a) Mekkora a rúd fajlagos alakváltozása?

b) Hogyan változik ez, ha megakadályozzuk a rúd keresztirányú alakváltozását?

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
F = 10 # kN
L = 160 # mm
d = 79.8 # mm
E = 200 # GPa
nu = 0.3

# Átváltás N, mm, MPa rendszerbe
F = F * 10**3 # N
E = E * 10**3 # MPa
```

##### A keresztmetszet területe

```python
A = d**2 * sp.pi / 4
display(Math(rf'A = \frac{{d^2 \pi}}{{4}} = {latex(A.evalf(6))} ~\mathrm{{mm^2}}'))
```

##### Feszültség $x$ irányban

```python
sigma_x = -F / A
display(Math(rf'\sigma_x = -\frac{{F}}{{A}} = {latex(sigma_x.evalf(3))} ~\mathrm{{MPa}}'))

# Feszültségi tenzor (egytengelyű feszültségi állapot)
sigma = sp.Matrix([[sigma_x, 0, 0],
                   [      0, 0, 0],
                   [      0, 0, 0]])
display(Math(rf'\boldsymbol{{\sigma}} = {latex(sigma.evalf(3))} ~\mathrm{{MPa}}'))
```

#### a) Szabad alakváltozás

A rúd keresztirányban szabadon alakváltozhat, így $\varepsilon_y$ és $\varepsilon_z$ nem zérus.

```python
# Hooke törvény egytengelyű feszültségi állapotban
eps_x = sigma_x / E
display(Math(rf'\varepsilon_x = \frac{{\sigma_x}}{{E}} = {latex(eps_x.evalf(3))}'))

# Keresztirányú alakváltozás
eps_y = -nu * eps_x
eps_z = -nu * eps_x
display(Math(rf'\varepsilon_y = \varepsilon_z = -\nu \cdot \varepsilon_x = {latex(eps_y.evalf(3))}'))
```

```python
# Alakváltozási tenzor
eps = sp.Matrix([[eps_x,     0,     0],
                 [    0, eps_y,     0],
                 [    0,     0, eps_z]])
display(Math(rf'\boldsymbol{{\varepsilon}} = {latex((eps * 10**6).evalf(3))} \cdot 10^{{-6}}'))
```

##### Fajlagos térfogatváltozás

```python
eps_V = eps.trace()
display(Math(rf'\varepsilon_V = \varepsilon_I = \mathrm{{tr}}(\boldsymbol{{\varepsilon}}) = \varepsilon_x + \varepsilon_y + \varepsilon_z = {latex(eps_V.evalf(3))}'))
```

#### b) Gátolt alakváltozás

A rúd keresztirányú alakváltozását megakadályozzuk ($\varepsilon_y = \varepsilon_z = 0$), így az alakváltozási tenzor:

```python
eps_x = sp.Symbol('varepsilon_x') # Ismeretlen fajlagos nyúlás x irányban

# Alakváltozási tenzor
eps = sp.Matrix([[eps_x, 0, 0],
                 [    0, 0, 0],
                 [    0, 0, 0]])
display(Math(rf'\boldsymbol{{\varepsilon}} = {latex(eps)}'))
```

##### Általános Hooke törvény

$$\boldsymbol{\sigma} = \frac{E}{1+\nu} \left( \boldsymbol{\varepsilon} + \frac{\nu}{1-2\nu} \cdot \varepsilon_I \cdot \mathbf{E} \right)$$

ahol:
- $\boldsymbol{\sigma}$: a feszültségi tenzor,
- $\boldsymbol{\varepsilon}$: az alakváltozási tenzor,
- $E$: a rugalmassági modulusz,
- $\nu$: a Poisson-tényező,
- $\varepsilon_I = \mathrm{tr}(\boldsymbol{\varepsilon})$: az alakváltozási tenzor első skalárinvariánsa (a főátlóbeli elemek összege),
- $\mathbf{E}$: az egységmátrix.

```python
I = sp.eye(3) # Egységmátrix

# Az alakváltozási tenzor első skalárinvariánsa
eps_I = eps.trace()
display(Math(rf'\varepsilon_I = \mathrm{{tr}}(\boldsymbol{{\varepsilon}}) = {latex(eps_I)}'))
```

```python
sigma_Hooke = E / (1 + nu) * (eps + nu / (1 - 2*nu) * eps_I * I)

# Főátló 1. eleme (x irány)
display(Math(rf'\sigma_x = \frac{{E}}{{1+\nu}} \left( \varepsilon_x + \frac{{\nu}}{{1-2\nu}} \cdot \varepsilon_x \right) = \frac{{E (1-\nu)}}{{(1+\nu)(1-2\nu)}} \cdot \varepsilon_x = {latex(sigma_Hooke[0, 0].evalf(6))}'))
```

##### $\sigma_x$ ismert a terhelésből

```python
eq = sigma_Hooke[0, 0] - sigma_x

# Rendszer megoldása
sol = sp.solve(eq, eps_x)
eps_x_sol = sol[0]
display(Math(rf'\varepsilon_x = \frac{{(1+\nu)(1-2\nu)}}{{E (1-\nu)}} \cdot \sigma_x = {latex(eps_x_sol.evalf(3))}'))
```

##### Fajlagos térfogatváltozás

```python
eps_V = eps_I.subs(eps_x, eps_x_sol)
display(Math(rf'\varepsilon_V = \varepsilon_I = \mathrm{{tr}}(\boldsymbol{{\varepsilon}}) = \varepsilon_x = {latex(eps_V.evalf(3))}'))
```

A keresztirányú alakváltozás megakadályozásával a fajlagos térfogatváltozás abszolút értéke megnő, mivel a rúd keresztirányban nem tud tágulni.
