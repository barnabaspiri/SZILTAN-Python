##### Feladat (Példatár 1.12)

Mekkora lehet a megoszló terhelés $p$ intenzitása (a tartó közepén), ha az anyagra megengedett feszültség $σ_\mathrm{meg} = 50 ~\mathrm{MPa}$ ? A tartó mérete ismert: $b = 10 ~\mathrm{cm}$ .

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
L = 6 # m
b = 10 # cm
sigma_meg = 50 # MPa

# Átváltás N, mm, MPa rendszerbe
L = L * 10**3 # mm
b = b * 10 # mm
```

##### Szabadtest ábra

<img src="img/02.jpg" alt="Feladat" width=300px>

##### A reakciók

```python
A_x, A_z, B_z = sp.symbols('A_x, A_z, B_z')
x = sp.Symbol('x', positive=True)
p = sp.Symbol('p', positive=True)
```

##### Egyensúlyi egyenletek

```python
# x irányú erő egyensúly
eq1 = A_x
display(Math(rf'\sum F_x = 0 \quad \longrightarrow \quad {latex(eq1)} = 0'))

# y irányú erő egyensúly
eq2 = A_z - p*2*L/2 + B_z
display(Math(rf'\sum F_y = 0 \quad \longrightarrow \quad {latex(eq2)} = 0'))

# z irányú nyomatéki egyensúly
eq3 = -p*2*L * L/2 + B_z*2*L
display(Math(rf'\sum M_A = 0 \quad \longrightarrow \quad {latex(eq3)} = 0'))
```

##### Egyensúlyi egyenletek megoldása

```python
sol = sp.solve([eq1, eq2, eq3], [A_x, A_z, B_z])

A_xsol = sol[A_x]
display(Math(rf'A_x = {latex(A_xsol)}'))

A_zsol = sol[A_z]
display(Math(rf'A_z = {latex(A_zsol)}'))

B_zsol = sol[B_z]
display(Math(rf'B_z = {latex(B_zsol)}'))
```

##### Hajlítónyomatéki igénybevétel

```python
# Megoszló helyettesítése 'x' koordináta esetén
p_x = p/L*x

# Hajlítónyomatéki függvény
Mh_I = -A_zsol*x + p_x*x/2 * x/3
display(Math(rf'M_{{h,I}} = {latex(Mh_I)}'))
```

##### Maximális hajlítónyomaték

```python
# Először szélsőértékek Abs nélkül, majd Max
M_min = sp.minimum(Mh_I, x, sp.Interval(0, L))
M_max = sp.maximum(Mh_I, x, sp.Interval(0, L))

Mh_Imax = sp.Max(sp.Abs(M_min), sp.Abs(M_max))
display(Math(rf'M_{{h,Imax}} = {latex(Mh_Imax)}'))
```

##### Az ébredő feszültség

```python
# Keresztmetszeti tényező
K = b*(2*b)**2 / 6

# Az ébredő feszültség
sigma_max = Mh_Imax / K
```

##### Határhelyzetben $\sigma_\mathrm{max} = \sigma_\mathrm{meg}$

```python
eq = sigma_max - sigma_meg

sol = sp.solve(eq, p)
psol = sol[0]

display(Math(rf'p_{{max}} = {latex(round(psol, 2))} ~ \mathrm{{N/mm}} = {latex(round(psol, 2))} ~ \mathrm{{kN/m}}'))
```
