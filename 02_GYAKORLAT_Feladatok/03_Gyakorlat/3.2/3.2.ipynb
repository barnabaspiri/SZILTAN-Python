##### Feladat (Példatár 1.8)

Egy függőleges oszlop terhelése a felső végén működő koncentrált erő az ábrán látható módon. Célunk meghatározni az oszlop anyagát és keresztmetszetét az alábbi módon:

a) Tömör kör keresztmetszetű fából kívánjuk elkészíteni.

b) Alumínium csőből gyártjuk le úgy, hogy a cső falvastagsága a külső átmérő nyolcada.

A fára és az alumíniumra megengedhető feszültségek: $σ_\mathrm{meg,fa} = 15 ~\mathrm{MPa}$ , $σ_\mathrm{meg,alu} = 50 ~\mathrm{MPa}$ . Határozzuk meg a szükséges méreteket!

##### Csomagok importálása

```python
import numpy as np # Numerikus számítások
import matplotlib.pyplot as plt # Ábrázolás
plt.rcParams['mathtext.fontset'] = 'cm' # Betűtípus az ábrákon
plt.rcParams['font.family'] = 'serif' # Betűtípus az ábrákon
plt.rcParams['font.serif'] = ['cmr10'] # Betűtípus az ábrákon
plt.rcParams['axes.formatter.use_mathtext'] = True # Betűtípus az ábrákon
plt.rcParams['axes.unicode_minus'] = False # Betűtípus az ábrákon
from IPython.display import display, Math, Latex # Szép kimenet Jupyter notebookban
import sympy as sp # Szimbólikus számítások
from sympy import latex # Szép kimenet Jupyter notebookban
sp.init_printing() # Szép kimenet Jupyter notebookban
```

##### Adatok

```python
h = 2.5 # m
P = 12 # kN
sigma_meg_fa = 15 # MPa
sigma_meg_alu = 50 # MPa

# Átváltás N, mm, MPa rendszerbe
P = P * 10**3 # N
h = h * 10**3 # mm
```

##### Maximális hajlítónyomaték

```python
Mh_max = P * h
display(Math(rf'M_{{h,max}} = {latex(round(Mh_max,3))} ~\mathrm{{Nmm}}'))
```

##### a) Tömör fa keresztmetszet

```python
# Másodrendű nyomaték
d_1 = sp.Symbol('d_1')
I_yfa = d_1**4*sp.pi / 64

# Az ébredő feszültség a hajlítás hatására 
sigma_fa = Mh_max / I_yfa * (d_1/2)

# Határhelyzetben egyenlő a megengedhető feszültséggel
eq = sigma_fa - sigma_meg_fa

# Rendszer megoldása
sol = sp.solve(eq, d_1)
d_1 = sol[0]

display(Math(rf'd_{{1}} = {latex(round(d_1,3))} ~\mathrm{{mm}}'))
```

##### b) Cső alumínium keresztmetszet

```python
# Másodrendű nyomaték
d_2 = sp.Symbol('d_1')
d_2i = 3/4*d_2
I_yalu = (d_2**4 - d_2i**4)*sp.pi / 64

# Az ébredő feszültség a hajlítás hatására 
sigma_alu = Mh_max / I_yalu * (d_2/2)

# Határhelyzetben egyenlő a megengedhető feszültséggel
eq = sigma_alu - sigma_meg_alu

# Rendszer megoldása
sol = sp.solve(eq, d_2)
d_2 = sol[0]

display(Math(rf'd_{{2}} = {latex(round(d_2,3))} ~\mathrm{{mm}}'))
```
